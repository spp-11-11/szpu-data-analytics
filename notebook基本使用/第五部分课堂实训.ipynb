```python
# 离差标准化数据
# 离差标准化函数
def max_min_scale(data):
    return (data-data.min())/(data.max()-data.min())
```

```python
# 标准差标准化数据
def std_scale(data):
    return (data-data.mean())/data.std()
```

```python
# 小数定标标准化数据
import numpy as np
##自定义小数定标标准化函数
def decimal_scale(data):
    return data/10**np.ceil(np.log10(data.abs().max()))
```
