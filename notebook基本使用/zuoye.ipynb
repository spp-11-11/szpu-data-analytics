```python
#240200341 赵书鹏
#导入库
import numpy as np

#输入n
n = int(input("请输入一个正整数："))

#创建空数组，然后循环输入，最后输出
arr = np.array([])
for i in range(n):
    arr = np.append(arr,0)
    arr = np.append(arr,n)
print(arr)
```
