## 2.1 掌握 NumPy 数组对象 ndarray
Python 提供了一个array 模块。array 和list 不同，它直接保存数值，和C 语言的一维
数组比较类似。但是由于Python 的array 模块不支持多维，也没有各种运算函数，因此也
不适合做数值运算。NumPy 的诞生弥补了这些不足。NumPy 提供了一种存储单一数据类型
的多维数组——ndarray。

```python
# Python的列表
a = [1,2,3]
b = [4,5,6]
print(a+b)
print(a*3)
#print(a+2)
# python的列表并不是c的数组,列表就是一个容器无法参与计算，但是数组可以参与计算
```

### 2.1.2 生成随机数

![](img/创建随机数组.png)  

```python
# pythond的标准库random
import random
random.random()
    #生产0-1之间的随机数
random.randint(1,3)
    #生产1-3之间的随机整数（包含结束值）（无法生成浮点数）
random.unifrom(2,4)
    #生成2-4之间的随机浮点数（包含结束值）
```

```python
#代码2-14 
#生成随机数组（注意是Numpy库的random子库中的random函数）
import numpy as np
np.random.random(5)
np.random.random((3,4))
    #用Numpy库的random子库中的random函数只能接受一个参数，所以二维数组要用元组的形式
```

```python
#代码2-15
#生成服从均匀分布的随机数数组（注意是Numpy库的random子库中的rand函数）
np.random.rand(5)
np.random.rand(3,4)
    #而用Numpy库的random子库中的rand函数可以接受多个参数，所以不需要用元组的形式
```

```python
#代码2-16
#生成服从正态分布的随机数数组（注意是Numpy库的random子库中的randn函数）
np.random.randn(5)
np.random.randn(3,4)
```

```python
#代码2-17
#生成给定上下限的随机整数数组（注意是Numpy库的random子库中的randint函数）
np.random.randint(2,10,(3,4))
    #在2-10的范围生产3行4列的二维数组
```
